```python
import requests
import pandas as pd

# Koordinat area Kampus UI / Depok
LAT, LON = -6.3645, 106.8287

# 1. Uji API Kualitas Udara (Januari 2024)
r_air = requests.get(
    "https://air-quality-api.open-meteo.com/v1/air-quality",
    params={
        "latitude": LAT,
        "longitude": LON,
        "start_date": "2024-01-01",
        "end_date": "2024-01-31",
        "hourly": "pm2_5,pm10,us_aqi",
        "domains": "cams_global",
        "timezone": "Asia/Jakarta",
    },
    timeout=30,
)

print("Status Air Quality:", r_air.status_code)
if r_air.status_code == 200:
    df_air = pd.DataFrame(r_air.json()["hourly"])
    print(df_air.head(3))
    print("\nProporsi Data Kosong (NaN) Udara:\n", df_air.isna().mean())

# 2. Uji API Cuaca Historis
r_weather = requests.get(
    "https://archive-api.open-meteo.com/v1/archive",
    params={
        "latitude": LAT,
        "longitude": LON,
        "start_date": "2024-01-01",
        "end_date": "2024-01-31",
        "hourly": "temperature_2m,relative_humidity_2m,precipitation,rain,wind_speed_10m",
        "timezone": "Asia/Jakarta",
    },
    timeout=30,
)

print("\nStatus Weather:", r_weather.status_code)
if r_weather.status_code == 200:
    df_weather = pd.DataFrame(r_weather.json()["hourly"])
    print(df_weather.head(3))
    print("\nProporsi Data Kosong (NaN) Cuaca:\n", df_weather.isna().mean())
```
